```python
import torch
import numpy as np
from torch.utils.data import Dataset, DataLoader
```

```python
class MorphFormerDataset(Dataset):

    def __init__(self, base_dataset):
        self.base_dataset = base_dataset

    def __len__(self):
        return len(self.base_dataset)

    def __getitem__(self, idx):

        patch, label = self.base_dataset[idx]

        # SalinasPatchDataset gives:
        # [204, 15, 15]

        return (
            torch.from_numpy(
                patch.astype(np.float32)
            ),
            torch.tensor(
                label,
                dtype=torch.long,
            ),
        )
```

```python
import torch
import torch.nn as nn
import numpy as np

# Make sure the original repository uses our patch size
import model
model.patchsize = 15

morphformer_model = CNN(
    FM=16,
    NC=204,
    Classes=16,
    HSIOnly=True,
).to(device)

print(morphformer_model)
```

```python
morphformer_params = sum(
    p.numel()
    for p in morphformer_model.parameters()
    if p.requires_grad
)

print(
    f"Trainable parameters: {morphformer_params:,}"
)

print(
    f"Parameters (M): "
    f"{morphformer_params / 1e6:.6f}"
)
```

```python
optimizer = torch.optim.Adam(
    morphformer_model.parameters(),
    lr=5e-4,
    weight_decay=5e-3,
)

loss_func = nn.CrossEntropyLoss()

scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer,
    step_size=50,
    gamma=0.9,
)
```

```python
from sklearn.metrics import accuracy_score, f1_score
from tqdm.auto import tqdm

MORPH_EPOCHS = 50

best_val_macro_f1 = -1.0
best_epoch = 0
best_state = None

history = []

for epoch in range(1, MORPH_EPOCHS + 1):

    # -------------------------
    # Training
    # -------------------------
    morphformer_model.train()

    running_loss = 0.0
    n_train = 0

    train_bar = tqdm(
        morph_train_loader,
        desc=f"Epoch {epoch:02d}/{MORPH_EPOCHS}",
        leave=False,
    )

    for x, y in train_bar:

        x = x.to(
            device,
            dtype=torch.float32,
        )

        y = y.to(
            device,
            dtype=torch.long,
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = morphformer_model(x)

        loss = loss_func(
            logits,
            y,
        )

        loss.backward()

        optimizer.step()

        batch_size = y.size(0)

        running_loss += (
            loss.item() * batch_size
        )

        n_train += batch_size

    train_loss = (
        running_loss / n_train
    )

    # -------------------------
    # Validation
    # -------------------------
    morphformer_model.eval()

    val_true = []
    val_pred = []

    with torch.no_grad():

        for x, y in morph_val_loader:

            x = x.to(
                device,
                dtype=torch.float32,
            )

            logits = morphformer_model(x)

            preds = torch.argmax(
                logits,
                dim=1,
            )

            val_true.extend(
                y.numpy()
            )

            val_pred.extend(
                preds.cpu().numpy()
            )

    val_acc = accuracy_score(
        val_true,
        val_pred,
    )

    val_macro_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0,
    )

    scheduler.step()

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_macro_f1,
    })

    print(
        f"Epoch {epoch:02d}/{MORPH_EPOCHS} | "
        f"Loss: {train_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro-F1: {val_macro_f1:.4f}"
    )

    # -------------------------
    # Best checkpoint
    # -------------------------
    if val_macro_f1 > best_val_macro_f1:

        best_val_macro_f1 = val_macro_f1
        best_epoch = epoch

        best_state = {
            k: v.detach().cpu().clone()
            for k, v in morphformer_model.state_dict().items()
        }

print()
print("Best epoch:", best_epoch)
print(
    "Best validation Macro-F1:",
    best_val_macro_f1,
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
)

morphformer_model.eval()

all_true = []
all_pred = []

with torch.no_grad():
    for x, y in morph_test_loader:

        x = x.to(
            device,
            dtype=torch.float32,
        )

        logits = morphformer_model(x)

        preds = torch.argmax(
            logits,
            dim=1,
        )

        all_true.append(
            y.cpu().numpy()
        )

        all_pred.append(
            preds.cpu().numpy()
        )

y_true_morph = np.concatenate(all_true)
y_pred_morph = np.concatenate(all_pred)

# Salinas has 16 classes: 0-15
salinas_labels = np.arange(16)

morph_accuracy = accuracy_score(
    y_true_morph,
    y_pred_morph,
)

morph_macro_f1 = f1_score(
    y_true_morph,
    y_pred_morph,
    labels=salinas_labels,
    average="macro",
    zero_division=0,
)

print("=" * 60)
print("MORPHFORMER SALINAS TEST RESULTS")
print("=" * 60)
print(f"Accuracy : {morph_accuracy:.4f}")
print(f"Macro-F1 : {morph_macro_f1:.4f}")
```

```python
import time
import numpy as np

morphformer_model.eval()

benchmark_batch_size = 128

x_bench, _ = next(
    iter(morph_test_loader)
)

x_bench = x_bench[:benchmark_batch_size].to(
    device,
    dtype=torch.float32,
)

# Warm-up
with torch.no_grad():
    for _ in range(20):
        _ = morphformer_model(x_bench)

torch.cuda.synchronize()

# Measured runs
times_ms = []

with torch.no_grad():

    for _ in range(100):

        torch.cuda.synchronize()

        start = time.perf_counter()

        _ = morphformer_model(x_bench)

        torch.cuda.synchronize()

        end = time.perf_counter()

        times_ms.append(
            (end - start) * 1000
        )

times_ms = np.asarray(times_ms)

mean_batch_ms = times_ms.mean()
median_batch_ms = np.median(times_ms)
p95_batch_ms = np.percentile(
    times_ms,
    95,
)
std_batch_ms = times_ms.std()

per_sample_ms = (
    mean_batch_ms
    / benchmark_batch_size
)

throughput = (
    benchmark_batch_size
    / (mean_batch_ms / 1000)
)

print("=" * 60)
print("MORPHFORMER RTX 2050 LATENCY")
print("=" * 60)
print(f"Batch size : {benchmark_batch_size}")
print(f"Mean       : {mean_batch_ms:.3f} ms")
print(f"Median     : {median_batch_ms:.3f} ms")
print(f"P95        : {p95_batch_ms:.3f} ms")
print(f"Std        : {std_batch_ms:.3f} ms")
print(f"Per sample : {per_sample_ms:.4f} ms")
print(f"Throughput : {throughput:.2f} samples/s")
```
