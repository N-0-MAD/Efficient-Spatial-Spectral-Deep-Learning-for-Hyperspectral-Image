# 04 - Hybrid Spatial-Spectral Model

Goal: test the proposed architecture family: CNN spatial branch, spectral attention branch, and gated fusion.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

import torch
from dfm.models.hybrid import HybridSpatialSpectralClassifier
from dfm.models.transformer_baseline import IndependentBandTransformerClassifier
```

```python
batch_size = 4
channels = 64
patch_size = 32
num_classes = 5

x = torch.randn(batch_size, channels, patch_size, patch_size)

hybrid = HybridSpatialSpectralClassifier(
    in_channels=channels,
    num_classes=num_classes,
    feature_dim=128,
    fusion="gated",
)

logits = hybrid(x)
print("Hybrid logits:", logits.shape)
```

```python
transformer = IndependentBandTransformerClassifier(
    num_classes=num_classes,
    patch_size=16,
    embed_dim=128,
    depth=2,
    num_heads=4,
    max_bands=256,
)

baseline_logits = transformer(x)
print("Transformer logits:", baseline_logits.shape)
```
